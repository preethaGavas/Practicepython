```python
#Employee Directory: Create an Employee class with employee ID, name, designation and salary. Create multiple objects and
#display employees based on a given designation.
class Employee:
    def __init__(self,emp_id,name,desig,salary):
        self.emp_id=emp_id
        self.name=name
        self.desig=desig
        self.salary=salary
    def display(self):
        print(self.emp_id)
        print(self.name)
        print(self.desig)
        print(self.salary)
        print()
employees = [
    Employee(101, "Arun", "Developer", 45000),
    Employee(102, "Priya", "Tester", 40000),
    Employee(103, "Ravi", "Developer", 50000),
    Employee(104, "Meena", "HR", 42000),
    Employee(105, "Kiran", "Tester", 38000)
]

for emp in employees:
    if emp.desig == "Developer":
      emp.display()
```

```python
#Mobile Phone: Create a Mobile class and create three objects representing
#different phones. Display their details and
#compare whether two references point to the same object.
class Mobile:
  def __init__(self,name,price,color):
    self.name=name
    self.price=price
    self.color=color
  def display(self):
    print(self.name)
    print(self.price)
    print(self.color)
m1 = Mobile("Moto",20000,"Red")
m2 = Mobile("Samsung",20400,"Blue")
m3 = m2
print(id(m1))
print(id(m2))
print(id(m1)==id(m2))
print(id(m3)==id(m2))

```

```python
#Library Books: Create multiple Book objects and store them in a list.
#Search the list for a book by ISBN and demonstrate
#object identity versus equality.
class Books:
  def __init__(self,isbn,title,author):
    self.isbn=isbn
    self.title=title
    self.author=author
  def display(self):
    print(self.isbn)
    print(self.title)
    print(self.author)
book = [Books(1234,"Python","Preetha"),Books(1256,"Java","Preetha")]
for b in book:
  if b.isbn == 1234:
    b.display()

print(id(book[0])==id(book[1]))
book1=book[0]
print(id(book[0])==id(book1))
print(id(book[0]) is id(book1))
```

```python
#Shopping Cart Item: Create a Product object with name and price. Add quantity dynamically and calculate the total price.
class Shopping_cart:
  def __init__(self,name,price,quantity):
    self.name=name
    self.price=price
    self.quantity=quantity
  def display(self):
    print(self.name)
    print(self.price)
    print(self.quantity)
    print()

cart = []

while True:
  name = input("Enter the name of the product (or 'q' to quit): ")
  if name == 'q':
    break
  price = float(input("Enter price: "))
  quantity = int(input("Enter quantity: "))
  product = Shopping_cart(name, price, quantity)
  # Add quantity dynamically
  #setattr(cart, "quantity", quantity)
  cart.append(product)
  print()
print("Shopping Cart")
total = 0

for product in cart:
  product.display()
  total += product.price * product.quantity
print("Pay",total)




```

```python
#Student Record Manager: Create student objects and dynamically
#add attributes such as email and grade. Use getattr(), setattr(), hasattr() and delattr() to manage student information.
class Student:
  def __init__(self,name,Roll_num,dept):
    self.name=name
    self.Roll_num=Roll_num
    self.dept=dept
  def display(self):
    print(self.name)
    print(self.Roll_num)
    print(self.dept)
Student_Info = []

while True:
  name = input("Enter the name of the student (or 'q' to quit): ")
  if name == 'q':
    break
  Roll_num = int(input("Enter Roll_num: "))
  dept = input("Enter dept: ")
  student = Student(name,Roll_num,dept)
  email=input("Enter email: ")
  grade=input("Enter grade: ")
  setattr(student,"email",email)
  setattr(student,"Grade",grade)
  Student_Info.append(student)
  print()
print("Student Information")
for student in Student_Info:
    student.display()
    print("Email:",getattr(student,"email"))
    print("Grade:",getattr(student,"Grade"))
```

```python
#Bank Account: Create a BankAccount class with deposit(), withdraw() and get_balance() methods.
class BankAccount:
    def __init__(self, account_number, balance):
        self.account_number = account_number
        self.balance = balance
    def deposit(self,amount):
        self.balance+=amount
    def withdraw(self,amount):
        if amount>self.balance:
            print("Insufficient balance")
        else:
            self.balance-=amount
    def display(self):
        print("Account Number:",self.account_number)
        print("Balance:",self.balance)
account = BankAccount("123456789", 5000)
account.display()
account.deposit(1000)
account.display()
account.withdraw(2000)
#account.get_balance()
account.display()
```
