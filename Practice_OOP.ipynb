```python
#Student Profile: Create a Student class with name,
#roll number and department. Create an object and display the details.
class Student:
    def __init__(self,name,roll_num,dept):
        self.name=name
        self.roll_num=roll_num
        self.dept=dept
    def display(self):
        print(self.name)
        print(self.roll_num)
        print(self.dept)
s=Student("Preetha",1001,"SCOPE")
print("Student Details:")
s.display()
```

```python
#Employee Directory: Create an Employee class with employee ID, name, designation and salary. Create multiple objects and
#display employees based on a given designation.
class Employee:
    def __init__(self,emp_id,name,desig,salary):
        self.emp_id=emp_id
        self.name=name
        self.desig=desig
        self.salary=salary
    def display(self):
        print(self.emp_id)
        print(self.name)
        print(self.desig)
        print(self.salary)
        print()
employees = [
    Employee(101, "Arun", "Developer", 45000),
    Employee(102, "Priya", "Tester", 40000),
    Employee(103, "Ravi", "Developer", 50000),
    Employee(104, "Meena", "HR", 42000),
    Employee(105, "Kiran", "Tester", 38000)
]

for emp in employees:
    if emp.desig == "Developer":
      emp.display()
```

```python
#Mobile Phone: Create a Mobile class and create three objects representing
#different phones. Display their details and
#compare whether two references point to the same object.
class Mobile:
  def __init__(self,name,price,color):
    self.name=name
    self.price=price
    self.color=color
  def display(self):
    print(self.name)
    print(self.price)
    print(self.color)
m1 = Mobile("Moto",20000,"Red")
m2 = Mobile("Samsung",20400,"Blue")
m3 = m2
print(id(m1))
print(id(m2))
print(id(m1)==id(m2))
print(id(m3)==id(m2))

```

```python
#Library Books: Create multiple Book objects and store them in a list.
#Search the list for a book by ISBN and demonstrate
#object identity versus equality.
class Books:
  def __init__(self,isbn,title,author):
    self.isbn=isbn
    self.title=title
    self.author=author
  def display(self):
    print(self.isbn)
    print(self.title)
    print(self.author)
book = [Books(1234,"Python","Preetha"),Books(1256,"Java","Preetha")]
for b in book:
  if b.isbn == 1234:
    b.display()

print(id(book[0])==id(book[1]))
book1=book[0]
print(id(book[0])==id(book1))
print(id(book[0]) is id(book1))
```
