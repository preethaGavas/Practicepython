```python
#namespace as dict:
class Dog:
    def __init__(self, name):
        self.name = name

    def speak(self):
        print("Woof!")

dog = Dog("Tommy")

print(dog.__dict__)
print(Dog.__dict__)
```

```python
class Dog:
    """Represents a student."""
    def __init__(self, name):
        self.name = name
    """This is the constructor """
    def speak(self):
        print("Woof!")

dog = Dog("Tommy")

print(dog.__dict__)
print(Dog.__doc__)
```

```python
#Docstring class,method,constructor
class Dog:
    """Represents a Class Dog."""
    def __init__(self, name):
        """This is the constructor """
        self.name = name

    def speak(self):
        """This is the speak method"""
        print("Woof!")

dog = Dog("Tommy")

print(Dog.__doc__)
print(Dog.speak.__doc__)
print(Dog.__init__.__doc__)
```

```python
#type of a class and object
class Student:

    """This class represents a student."""

    def __init__(self, name, mark):
        """Initialize student details."""
        self.name = name
        self.mark = mark

    def display(self):
        """Display student details."""
        print(self.name, self.mark)


s1 = Student("Preetha", 95)
print(type(Student))
print(type(s1))
print(dir(s1))
```

| Feature                         | `type()`                               | `dir()`                                | `help()`                                              |
| ------------------------------- | -------------------------------------- | -------------------------------------- | ----------------------------------------------------- |
| **Purpose**                     | Identifies the type/class of an object | Lists available attributes and methods | Displays documentation about an object/class/function |
| **Syntax**                      | `type(object)`                         | `dir(object)`                          | `help(object)`                                        |
| **Returns**                     | Type/class information                 | List of names                          | Detailed documentation                                |
| **Main question answered**      | **"What is it?"**                      | **"What does it contain?"**            | **"How does it work?"**                               |
| **Used with classes**           | Finds the type of a class/object       | Shows class attributes and methods     | Shows class documentation and methods                 |
| **Used with objects**           | Finds object's class                   | Shows object's attributes/methods      | Shows object's documentation                          |
| **Shows user-defined methods?** | ❌ No                                   | ✅ Yes                                  | ✅ Yes                                                 |
| **Shows attributes?**           | ❌ No                                   | ✅ Yes                                  | Partially, depending on object                        |
| **Shows docstrings?**           | ❌ No                                   | Shows `__doc__` name                   | ✅ Yes                                                 |
| **Shows special methods?**      | ❌ No                                   | ✅ Yes, e.g. `__init__`, `__str__`      | ✅ Yes                                                 |
| **Output type**                 | `<class ...>`                          | List                                   | Documentation text                                    |
| **Example**                     | `type(s1)`                             | `dir(s1)`                              | `help(Student)`                                       |
| **Typical use**                 | Identify an object's class             | Explore an object's capabilities       | Learn how to use it                                   |


```python
#object identity
class Student:
    def __init__(self, name, mark):
        self.name = name
        self.mark = mark
        #print(self.name, self.mark) also works
    def display(self):
        print(self.name, self.mark)


s1 = Student("Preetha", 95)
s2 = Student("Dikshit", 100)
s1.display()
s2.display()
print(s1)
print(s2)
print(id(s1))
print(id(s2))
```

```python
#isinstance() issubclass()
print(isinstance(s1, Student))
print(issubclass(Student, object))
```

```python
#Class Attribute,InstanceAttribute,local variable
class BankAccount:
    # Class Attribute
    bank_name = "ABC Bank"
    def __init__(self, account_holder, balance):
        # Instance Attributes
        self.account_holder = account_holder
        self.balance = balance
    def deposit(self, amount):
        # Local Variable
        new_balance = self.balance + amount
        # Updating the instance attribute
        self.balance = new_balance
        print("Deposited Amount:", amount)
        print("New Balance:", self.balance)
# Creating objects
account1 = BankAccount("Preetha", 10000)
account2 = BankAccount("Anu", 15000)
# Accessing class attribute
print("Bank:", BankAccount.bank_name)
# Accessing instance attributes
print("Account Holder:", account1.account_holder)
print("Balance:", account1.balance)
print()
print("Account Holder:", account2.account_holder)
print("Balance:", account2.balance)
print()
# Calling method
account1.deposit(2000)
```

```python
#Dunder method
#__str__  => method defines how an object should be represented as a readable string.
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
       return f"{self.name} is {self.age} years old"


s = Student("Arun", 20)
print(str(s))
```

```python
#The __len__() method tells Python how long your object is.
#Can be used in Inventory,Adding Books in Library
class ShoppingCart:
    def __init__(self):
        self.items = []

    def add_item(self, item):
        self.items.append(item)

    def __len__(self):
        return len(self.items)


cart = ShoppingCart()

cart.add_item("Laptop")
cart.add_item("Mouse")
cart.add_item("Keyboard")

print(len(cart))
```

```python
#__add__ => Add the different objects of the same class may be to a new object
class ShoppingCart:
    def __init__(self, items):
        self.items = items

    def __add__(self, other):
        return ShoppingCart(self.items + other.items)


cart1 = ShoppingCart(["Laptop", "Mouse"])
cart2 = ShoppingCart(["Keyboard", "Headset"])

cart3 = cart1 + cart2

print(cart3.items)
```

```python
#vars() => see the attributes stored in an object or the variables in a namespace.
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

s = Student("Arun", 20)

print(vars(s))
```

```python
#validation inside __init__

class BankAccount:
    def __init__(self, account_no, balance):
        if balance < 0:
            raise ValueError("Balance cannot be negative")

        self.account_no = account_no
        self.balance = balance

a1 = BankAccount(101, 5000)   # Valid
print(vars(a1))
#a2 = BankAccount(102, -100)   # Invalid - this object will not be created since negative
```

```python
#multiple attributes initialization in constructor
class Student:
    def __init__(self, name, age, department, marks):
        self.name = name
        self.age = age
        self.department = department
        self.marks = marks

s1 = Student("Arun", 20, "CSE", 85)

print(s1.name)
print(s1.age)
print(s1.department)
print(s1.marks)
```

```python
#constructor chaining is possible without inheritance
class Student:
    def __init__(self, name, age):
        self.set_basic_details(name, age)
        self.marks = 85
        print(self.marks)
    def set_basic_details(self, name, age):
        self.name = name
        self.age = age
s = Student("Arun", 20)
print(s.name)
print(s.age)
print(s.marks)
```

```python
#mutable Class Attributes
# key point:Avoid mutable class attributes when each object needs its own separate data.
class Student:
    subjects = []   # Mutable class attribute

    def __init__(self, name):
        self.name = name

s1 = Student("Arun")
s2 = Student("Priya")

s1.subjects.append("Python")
s1.subjects.append("Java")
s2.subjects.append("C++")

print(s1.subjects)
print(Student.subjects)
```

```python
#method Overriding
class Vehicle:
    def start(self):
        print("Vehicle starts")


class Car(Vehicle):
    def start(self):
        print("Car starts with a key")


class Bike(Vehicle):
    def start(self):
        print("Bike starts with a self-start button")
v = Bike()
v.start()
```

```python
#method overloading using *args
class Calculator:

    def add(self, *numbers):
        return sum(numbers)

calc = Calculator()

print(calc.add(10, 20))
print(calc.add(10, 20, 30))
print(calc.add(10, 20, 30, 40))
```

```python
#method overloading using default arguments
class Calculator:

    def add(self, a, b, c=0):
        return a + b + c


calc = Calculator()

print(calc.add(10, 20))
print(calc.add(10, 20, 30))
```

```python
#polymorphism - objects of the Classes in list
class Animal:
    def speak(self):
        print("Animal sound")
class Dog(Animal):
    def speak(self):
        print("Dog barks")
class Cat(Animal):
    def speak(self):
        print("Cat meows")
animals = [Dog(),Cat(),Animal()]
for animal in animals:
    animal.speak()
```
