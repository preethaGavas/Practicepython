```python
#namespace as dict:
class Dog:
    def __init__(self, name):
        self.name = name

    def speak(self):
        print("Woof!")

dog = Dog("Tommy")

print(dog.__dict__)
print(Dog.__dict__)
```

```python
class Dog:
    """Represents a student."""
    def __init__(self, name):
        self.name = name
    """This is the constructor """
    def speak(self):
        print("Woof!")

dog = Dog("Tommy")

print(dog.__dict__)
print(Dog.__doc__)
```

```python
#Docstring class,method,constructor
class Dog:
    """Represents a student."""
    def __init__(self, name):
        """This is the constructor """
        self.name = name

    def speak(self):
        """This is the speak method"""
        print("Woof!")

dog = Dog("Tommy")

print(Dog.__doc__)
print(Dog.speak.__doc__)
print(Dog.__init__.__doc__)
```

```python
#type of a class and object
class Student:

    """This class represents a student."""

    def __init__(self, name, mark):
        """Initialize student details."""
        self.name = name
        self.mark = mark

    def display(self):
        """Display student details."""
        print(self.name, self.mark)


s1 = Student("Preetha", 95)
print(type(Student))
print(type(s1))
print(dir(s1))
```

```python
print(dir(s1))
```

```python
print(help(Student))
```

```python
print(help(s1))
```

| Feature                         | `type()`                               | `dir()`                                | `help()`                                              |
| ------------------------------- | -------------------------------------- | -------------------------------------- | ----------------------------------------------------- |
| **Purpose**                     | Identifies the type/class of an object | Lists available attributes and methods | Displays documentation about an object/class/function |
| **Syntax**                      | `type(object)`                         | `dir(object)`                          | `help(object)`                                        |
| **Returns**                     | Type/class information                 | List of names                          | Detailed documentation                                |
| **Main question answered**      | **"What is it?"**                      | **"What does it contain?"**            | **"How does it work?"**                               |
| **Used with classes**           | Finds the type of a class/object       | Shows class attributes and methods     | Shows class documentation and methods                 |
| **Used with objects**           | Finds object's class                   | Shows object's attributes/methods      | Shows object's documentation                          |
| **Shows user-defined methods?** | ❌ No                                   | ✅ Yes                                  | ✅ Yes                                                 |
| **Shows attributes?**           | ❌ No                                   | ✅ Yes                                  | Partially, depending on object                        |
| **Shows docstrings?**           | ❌ No                                   | Shows `__doc__` name                   | ✅ Yes                                                 |
| **Shows special methods?**      | ❌ No                                   | ✅ Yes, e.g. `__init__`, `__str__`      | ✅ Yes                                                 |
| **Output type**                 | `<class ...>`                          | List                                   | Documentation text                                    |
| **Example**                     | `type(s1)`                             | `dir(s1)`                              | `help(Student)`                                       |
| **Typical use**                 | Identify an object's class             | Explore an object's capabilities       | Learn how to use it                                   |


```python
#object identity
class Student:
    def __init__(self, name, mark):
        self.name = name
        self.mark = mark
        #print(self.name, self.mark) also works
    def display(self):
        print(self.name, self.mark)


s1 = Student("Preetha", 95)
s2 = Student("Dikshit", 100)
s1.display()
s2.display()
print(s1)
print(s2)
print(id(s1))
print(id(s2))
```

```python
#isinstance() issubclass()
print(isinstance(s1, Student))
print(issubclass(Student, object))
```

```python
#object equality
s1 = s2
print(s1 is s2)
print(s1 == s2)
```

```python
s3 = Student("Lakshitha", 100)
print(s1 is s3)
print(s1 == s3)
```

```python
#Class Attribute,InstanceAttribute,local variable
class BankAccount:
    # Class Attribute
    bank_name = "ABC Bank"
    def __init__(self, account_holder, balance):
        # Instance Attributes
        self.account_holder = account_holder
        self.balance = balance
    def deposit(self, amount):
        # Local Variable
        new_balance = self.balance + amount
        # Updating the instance attribute
        self.balance = new_balance
        print("Deposited Amount:", amount)
        print("New Balance:", self.balance)
# Creating objects
account1 = BankAccount("Preetha", 10000)
account2 = BankAccount("Anu", 15000)
# Accessing class attribute
print("Bank:", BankAccount.bank_name)
# Accessing instance attributes
print("Account Holder:", account1.account_holder)
print("Balance:", account1.balance)
print()
print("Account Holder:", account2.account_holder)
print("Balance:", account2.balance)
print()
# Calling method
account1.deposit(2000)
```
