```python
print("Think of a number")
print("I will try to guess it!")

low = 1
high = 10000

while low <= high:
    guess = (low + high) // 2

    print("\nIs your number", guess, "?")
    response = input("Enter H for Higher, L for Lower, C for Correct: ").upper()

    if response == "C":
        print("I guessed your number! 🎉")
        print("Your number is:", guess)
        break

    elif response == "H":
        # Actual number is higher than the guess
        low = guess + 1

    elif response == "L":
        # Actual number is lower than the guess
        high = guess - 1

    else:
        print("Invalid input! Please enter H, L, or C.")

else:
    print("Something went wrong. Please check your answers.")
```

```python
def get_guess(low, high):
    """Get a valid guess from the user."""
    while True:
        guess = int(input(f"Guess a number between {low} and {high}: "))

        if low <= guess <= high:
            return guess
        else:
            print("Please enter a number within the given range.")


def guessing_game():
    """Play the number guessing game."""

    low = 1
    high = 100

    print("Think of a number between 1 and 100.")
    print("I will try to guess your number.")

    while low <= high:

        guess = get_guess(low, high)

        response = input(
            "Is your number (H)igher, (L)ower, or (C)orrect? ").lower()

        if response == "c":
            print(f"Great! Your number is {guess}.")
            break

        elif response == "h":
            # Number is higher than the guessed number
            low = guess + 1

        elif response == "l":
            # Number is lower than the guessed number
            high = guess - 1

        else:
            print("Please enter H, L, or C.")

    else:
        print("Something went wrong with the responses.")


# Function call
guessing_game()
```
